This cell imports the required Python libraries for data handling, visualization, model building, and evaluation.

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve
from imblearn.over_sampling import SMOTE
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.metrics import precision_recall_curve, average_precision_score
from sklearn.inspection import permutation_importance

```

The Breast Cancer Wisconsin dataset is used to distinguish between **malignant** and **benign** tumors based on the provided features.

The dataset contains **569** samples obtained from breast biopsy images, each described by **30** numerical features related to the shape and structure of the cell nucleus

This cell loads the Breast Cancer dataset directly from the UCI Machine Learning repository and stores it in a pandas DataFrame.

```python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/breast-cancer-wisconsin/wdbc.data'
data = pd.read_csv(url, header=None)
columns = ['ID','Diagnosis', 'radius1','texture1', 'perimeter1', 'area1', 'smoothness1', 'compactness1', 'concavity1', 'concave_points1','symmetry1','fractal_dimension1', 'radius2','texture2', 'perimeter2', 'area2',  'smoothness2', 'compactness2','concavity2', 'concave_points2', 'symmetry2','fractal_dimension2', 'radius3','texture3','perimeter3', 'area3', 'smoothness3', 'compactness3', 'concavity3', 'concave_points3', 'symmetry3', 'fractal_dimension3'   ]
data.columns = columns
data
```

This cell performs an initial exploration of the dataset by printing the first rows, checking the structure, and displaying the distribution of the target classes.

The dataset is composed of **357** benign samples and **212** malignant samples.

```python
print(data.head())
print(data.info())
print(data['Diagnosis'].value_counts())
```

This cell removes the non-informative "ID" column and encodes the target variable 'Diagnosis' (Malignant=1, Benign=0)


```python
data =data.drop(['ID'], axis=1)
data['Diagnosis']= data['Diagnosis'].map({'M': 1, 'B': 0})
```

This cell plots a **histogram** to visualize the imbalance between malignant and benign samples in the dataset.

```python
y = data['Diagnosis']
sns.countplot(x=y)
plt.title("Class Distribution Before SMOTE")
plt.xlabel("Diagnosis (0 = Benign, 1 = Malignant)")
plt.ylabel("Count")
plt.show()
```

Correlation Matrix Before vs After Feature Reduction

By comparing the correlation matrix before and after feature reduction, we can clearly observe the impact of removing highly correlated features:

- **Before reduction**: Several features exhibited very strong correlations (values close to 1), especially those derived from the same measurement group (e.g., `radius1`, `area1`, `perimeter1`). These redundant features can lead to overfitting and distort the learning process of certain models.
  
- **After reduction**: The most redundant features were removed using a correlation threshold (e.g., 0.95), resulting in a more compact and independent set of variables. The updated correlation matrix shows a more balanced structure, with fewer extreme correlation values.

This step improves the interpretability of the model and reduces the risk of multicollinearity, leading to potentially better generalization on unseen data.

```python
plt.figure(figsize=(12, 10))
sns.heatmap(data.corr(), annot=False, cmap='coolwarm', square=True)
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

This cell calculates the feature correlation matrix and automatically detects highly correlated features to remove redundancy.

```python
corr_matrix = data.corr().abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
to_drop = [column for column in upper.columns if any(upper[column] > 0.95)]
dataset = data.drop(columns=to_drop)
```

```python
# Visualize correlation matrix after dropping highly correlated features
plt.figure(figsize=(12, 10))
sns.heatmap(dataset.corr(), annot=False, cmap='coolwarm', square=True)
plt.title("Correlation Matrix After Feature Reduction")
plt.tight_layout()
plt.show()
```

This cell separates the dataset in two parts, the features (X) and the target variable (y), preparing the dataset for further analysis and model training.


```python
X = data.drop('Diagnosis', axis=1)
y = data['Diagnosis']
```

This cell standardizes all features to have zero mean and unit variance,
a crucial step for algorithms that are sensitive to feature scale.
It is important because different features may have very different value ranges,
and some algorithms can be biased toward features with larger magnitudes.


```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

Although many points appear as outliers in the boxplots, they often reflect the natural variability
and non-Gaussian distribution of the features. These values can still be useful for the model,
and no removal is required unless they are proven to be true anomalies.


```python
pd.DataFrame(X_scaled, columns=X.columns).plot(
    kind='box', subplots=True, layout=(5, 6), figsize=(15, 10), sharex=False, sharey=False)
plt.suptitle("Boxplots of Standardized Features")
plt.tight_layout()
plt.show()
```

This cell applies the SMOTE technique (Synthetic Minority Over-sampling Technique) to handle class imbalance by generating synthetic samples for the minority class.


```python
sm = SMOTE(random_state=42)
X_res, y_res = sm.fit_resample(X_scaled, y)
```

This cell splits the balanced dataset into training (70%) and testing (30%) sets to evaluate model performance on unseen data.

```python
X_train, X_test, y_train, y_test = train_test_split(X_res, y_res, test_size=0.3, stratify=y_res, random_state=42)
```

This cell defines the machine learning models to be tested: Random Forest, Logistic Regression, and K-Nearest Neighbors.


```python
models = {
    'Random Forest': RandomForestClassifier(random_state=42),
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'KNN': KNeighborsClassifier()
}
```

This cell trains and evaluates each model by generating predictions on the test set, calculating the AUC score, and applying 5-fold cross-validation to ensure robust performance.

```python
results = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    probs = model.predict_proba(X_test)[:, 1]
    auc = roc_auc_score(y_test, probs)
    cv_auc = cross_val_score(model, X_res, y_res, cv=5, scoring='roc_auc').mean()

    results[name] = {'model': model, 'auc': auc, 'cv_auc': cv_auc}

    print(f"\n=== {name} ===")
    print(confusion_matrix(y_test, preds))
    print(classification_report(y_test, preds))
    print(f"AUC: {auc:.3f} | CV AUC: {cv_auc:.3f}")





    precision, recall, _ = precision_recall_curve(y_test, probs, pos_label = 1)
    ap = average_precision_score(y_test, probs, pos_label = 1)

    plt.figure(figsize=(5,4))
    plt.plot(recall, precision, color='purple', lw=2, label=f'AP = {ap:.2f}')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.title(f'{name} - Precision-Recall Curve')
    plt.legend(loc='lower left')
    plt.show()



```

This cell generates ROC curves for all trained models, allowing a visual comparison of their classification performance and AUC scores.


```python
plt.figure(figsize=(8,6))
for name, res in results.items():
    fpr, tpr, _ = roc_curve(y_test, res['model'].predict_proba(X_test)[:, 1])
    plt.plot(fpr, tpr, label=f'{name} (AUC = {res["auc"]:.2f})')

plt.plot([0,1], [0,1], 'k--')
plt.title("ROC Curve Comparison")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.grid(True)
plt.show()
```

This cell generates and displays confusion matrices for each model, showing the distribution of correct and incorrect predictions across the two classes.


```python
for name, res in results.items():
    disp = ConfusionMatrixDisplay.from_estimator(res['model'], X_test, y_test)
    disp.ax_.set_title(f"Confusion Matrix - {name}")
    plt.grid(False)
    plt.show()
```

This cell extracts and plots the top 10 most important features according to the Random Forest model,enhancing interpretability.

```python
rf = results['Random Forest']['model']
importances = pd.Series(rf.feature_importances_, index=X.columns)
importances.nlargest(10).plot(kind='barh')
plt.title("Top 10 Feature Importances - Random Forest")
plt.xlabel("Importance")
plt.show()
```

The previous cell showed feature importances only for the Random Forest model, since it has a built-in `feature_importances_` attribute.  
For models like **KNN** and **Logistic Regression**, this metric is not available, so here we use **Permutation Importance** to estimate how much each feature affects model performance.


```python
models_to_check = {
    'KNN': results['KNN']['model'],
    'Logistic Regression': results['Logistic Regression']['model']
}

for name, model in models_to_check.items():
    print(f"\n🔹 Model: {name}")

    result = permutation_importance(model, X_test, y_test, n_repeats=10, random_state=42)

    importances = pd.Series(result.importances_mean, index=X.columns)

    importances.nlargest(10).plot(kind='barh')
    plt.title(f"Top 10 Feature Importances ({name} - Permutation)")
    plt.xlabel("Importance")
    plt.show()
```

This cell creates a summary table comparing the AUC scores and cross-validation results of all trained models.

```python
summary_df = pd.DataFrame({
    name: {
        'AUC': res['auc'],
        'CV AUC': res['cv_auc']
    } for name, res in results.items()
}).T

print("\n\n Confronto Modelli:")
print(summary_df)
```

This cell performs hyperparameter tuning on the Logistic Regression model using GridSearchCV to optimize its parameters for maximizing AUC performance.


Other models such as Random Forest and KNN were used with default parameters, as they generally perform well without extensive tuning.

```python
param_grid = {
    'C': [0.01, 0.1, 1, 10],
    'penalty': ['l2'],
    'solver': ['lbfgs']
}


grid = GridSearchCV(LogisticRegression(max_iter=1000), param_grid, cv=5, scoring='roc_auc')
grid.fit(X_train, y_train)
print("Best parameters for Logistic Regression:", grid.best_params_)
```

This cell identifies and prints the best-performing model based on the AUC score, concluding the analysis.



```python
best_model = max(results.items(), key=lambda x: x[1]['auc'])
print(f"\n The best model is: {best_model[0]} with AUC = {best_model[1]['auc']:.3f}")
```
